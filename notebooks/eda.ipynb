```python
# IMPORTS
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# HELPER FUNCTIONS

# Data loader
def load_data(file_path):
  df = pd.read_csv(file_path)
  df.columns = df.columns.str.lower()
  return df

# EDA: Shape and schema
def eda(df, df_name):
  print("\n")
  print(f"Showing EDA for {df_name} dataset \n")
  print(f"Shape: {df.shape}\n")
  print(f"Schema:\n{df.dtypes}\n")
  print(f"Columns:\n{df.columns}\n")
  print(f"Missing values:\n{df.isnull().sum()}\n")
  print(f"Unique states: {df['state'].nunique()}")
  print(f"Unique districts: {df['district'].nunique()}")
  print(f"Unique pincodes: {df['pincode'].nunique()}\n")
  print(f"Invalid entries: {(df['valid_flag'] == 'Not Valid').sum()}")
  print('-' * 40)
```

```python
# INITIALIZING DATAFRAMES

# Biometric data
file_path = 'data/processed/api_data_aadhar_biometric/api_data_aadhar_biometric_0_500000_cleaned.csv'
df_biometric_data = load_data(file_path)

# Demographic data
file_path = 'data/processed/api_data_aadhar_demographic/api_data_aadhar_demographic_0_500000_cleaned.csv'
df_demographic_data = load_data(file_path)

# Enrollment data
file_path = 'data/processed/api_data_aadhar_enrolment/api_data_aadhar_enrolment_0_500000_cleaned.csv'
df_enrolment_data = load_data(file_path)

# DATASETS EDA

eda(df_biometric_data, 'biometric_0_500000_cleaned')
eda(df_demographic_data, 'demographic_0_500000_cleaned')
eda(df_enrolment_data, 'enrolment_0_500000_cleaned')
```
